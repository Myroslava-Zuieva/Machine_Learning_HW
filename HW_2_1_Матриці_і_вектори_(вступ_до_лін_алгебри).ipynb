3. Задана матриця `D` розмірністю 2х2. Подумайте спочатку, яку розмірність Ви отрмаєте, коли домножите `result_matrix` на матрицю `D` і який результат буде. А потім виконайте множення і виведіть результат на екран.

```python
A = np.array(
    [[0.18, 0.53],
     [0.75, 0.47],
     [0.35, 0.21]])

B = np.array(
    [[0.17, 0.58, 0.75],
     [0.38, 0.11, 0.15]])

C = np.array([[0], [1]])

D = np.array([[0,1], [1,0]])

display(A, B, C, D)
```

```python
result_matrix = np.dot(B, A)

print("result_matrix:")
print(result_matrix)
print("Форма result_matrix:", result_matrix.shape)

result_vector = np.dot(result_matrix, C)
print("\nМноження result_matrix та C:")
print(result_vector)
print("Форма:", result_vector.shape)

result_matrix_D = np.dot(result_matrix, D)
print("\nМноження result_matrix та D:")
print(result_matrix_D)
print("Форма:", result_matrix_D.shape)
```
