### Домашні завдання на відпрацювання поняття функції та її похідної

1. Задана функція залежності оцінки від кількості годин, витрачених на навчання: $ f(x) = 10x - 5 $.
  1. З допомогою Python обчисліть похідну цієї функції.
  2. Використовуйте бібліотеку `matplotlib` для побудови графіків функції та її похідної в діапазоні значень від 0 до 10 годин. Проаналізуйте, як зміна кількості годин вивчення впливає на оцінку.



```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp

x = sp.symbols('x')
f = 10*x - 5
f_prime = sp.diff(f, x)

print("Функція:", f)
print("Похідна:", f_prime)

x_vals = np.linspace(0, 10, 100)
y_vals = 10*x_vals - 5
y_prime_vals = np.full_like(x_vals, 10)

plt.plot(x_vals, y_vals, label='f(x) = 10x - 5')
plt.plot(x_vals, y_prime_vals, label="f'(x) = 10")
plt.xlabel('Години навчання')
plt.ylabel('Значення')
plt.title('Функція та її похідна')
plt.legend()
plt.grid()
plt.show()
```

2. Використовуючи Python і `SymPy`, обчисліть похідні для функцій $ f(x) = \sin(x) $ та $ f(x) = \ln(x) $. Зобразіть ці функції та їхні похідні за допомогою `matplotlib`.



```python
x = sp.symbols('x', positive=True)

f_sin = sp.sin(x)
f_log = sp.log(x)

f_sin_prime = sp.diff(f_sin, x)
f_log_prime = sp.diff(f_log, x)

print(f"Функція синусу = {f_sin}, похідна = {f_sin_prime}")
print(f"Функція логарифму = {f_log}, похідна = {f_log_prime}")

xs_sin = np.linspace(0, 10, 100)
xs_log = np.linspace(0.1, 10, 100)

f1 = np.sin(xs_sin)
df_1 = np.cos(xs_log)

f2 = np.log(xs_log)
df_2 = 1 / xs_log

plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.plot(xs_sin, f1, label='sin(x)')
plt.plot(xs_sin, df_1, label="cos(x)")
plt.title('Функція sin(x) та її похідна')
plt.xlabel('x')
plt.ylabel('Значення')
plt.legend()
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(xs_log, f2, label='ln(x)')
plt.plot(xs_log, df_2, label="1/x")
plt.title('Функція ln(x) та її похідна')
plt.xlabel('x')
plt.ylabel('Значення')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```


3. Загальні витрати компанії від обсягу виробництва задані функцією $ f(x) = x^2 - 5x + 1 $, де $ x $ — кількість вироблених одиниць. Обчисліть похідну цієї функції з допомогою SymPy та візуалізуйте значення функції і її похідної на відрізку від 0 до 100.

```python
x = sp.symbols('x')

func = x**2 - 5*x + 1
f_prime = sp.diff(f, x)

print(f"Функція: {func}")
print(f"Похідна: {f_prime}")

xs = np.linspace(0, 100, 400)

y_vals = xs**2 - 5*xs + 1
y_prime_vals = 2*xs - 5

plt.figure(figsize=(10, 6))
plt.plot(xs, y_vals, label='f(x) = x^2 - 5x + 1')
plt.plot(xs, y_prime_vals, label="f'(x) = 2x - 5")
plt.xlabel('Кількість вироблених одиниць (x)')
plt.ylabel('Витрати / Похідна')
plt.title('Функція загальних витрат і її похідна')
plt.legend()
plt.grid(True)
plt.show()
```
